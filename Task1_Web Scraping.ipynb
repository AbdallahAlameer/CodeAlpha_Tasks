```python
import requests
from bs4 import BeautifulSoup
import json
```

```python

serp_Api_key = '57d6575e6264ec5ba4ab10adf5b6e05c01e6531e35b9e9804d9d1202ec6145c9'
parameters = {
    'api_key': serp_Api_key,
    'k': 'laptop',  
    'engine': 'amazon',
    's' : 'price-desc-rank'
}

search = requests.get('https://serpapi.com/search?engine=amazon', params=parameters)
response = search.json()
#print(json.dumps(response, indent=2))
```

```python
page =2 
all_organic_results = response.get('organic_results', [])
while True :
    parameters['page'] = page
    search = requests.get('https://serpapi.com/search?engine=amazon', params=parameters)
    response = search.json()
    
    if 'organic_results' not in response or not response['organic_results']:
        break
    organic_results = response['organic_results']
    all_organic_results.extend(organic_results)
    print(f'Page {page} results added. Total results: {len(all_organic_results)}')
    page += 1

    print(f'total results: {len(all_organic_results)}')
```

```python
import csv 

with open('amazon_laptops.csv', 'w', newline='', encoding='utf-8') as csvfile:
    fieldnames = ['Title', 'Brand', 'ASIN', 'Thumbnail', 'Price',  'Rating', 'Display Size', 'operating_system', 'ram', 'disk_size', 'Reviews', 'Link']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    for result in all_organic_results:
                writer.writerow({
                'Title': result.get('title'),
                'Brand': result.get('title').split()[0] if result.get('title') else 'Unknown',
                'ASIN': result.get('asin'),
                'Thumbnail': result.get('thumbnail'),
                'Price': result.get('price'),
                'Rating': result.get('rating'),
                'Display Size': result.get('specs', {}).get('display_size'),
                'operating_system': result.get('specs', {}).get('operating_system'),
                'ram': result.get('specs', {}).get('ram'),
                'disk_size' : result.get('specs', {}).get('disk_size'),
                'Reviews': result.get('reviews'),
                'Link': result.get('link_clean')
                                    }) 

print('Data has been written to amazon_laptops.csv')

 
```
