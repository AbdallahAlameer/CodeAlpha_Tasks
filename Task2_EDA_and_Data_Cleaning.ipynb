```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('amazon_laptops.csv')
```

## 1. Data Cleaning

```python
data.head(10)
```

```python
data.info()
```

```python
drop_columns = ['ASIN','Link','Thumbnail']
data.drop(drop_columns, axis=1, inplace=True)
```

```python
data['Price']=data['Price'].str.replace('$','').str.replace(',','').astype(float)
 
```

```python
data['display_size'] = data['Display Size'].str.replace(' inches','').astype(float)
data.drop('Display Size', axis=1, inplace=True)
```

```python
data['disk_size'].value_counts()
```

```python
def clean_disk_size(x):
    x = str(x).lower().strip()

    if x =='-' or x =='0 gb':
        return 0 
    try:
        num = float(x.split()[0])
    except:
        return 0
    
    if 'tb' in x : 
        res = num * 1024
    else:
        res = num
    
    if res > 50000 : 
        res = res/1024

    return res
     
```

```python
data['disk_size'] = data['disk_size'].apply(clean_disk_size)
```

```python
data.info()
```

```python
data['Price']= data['Price'].dropna()
data['Reviews'] =data['Reviews'].fillna(0).astype(int)  
data['ram'] = data['ram'].str.replace(' GB','')
data['ram'] = pd.to_numeric(data['ram'], errors='coerce')
```

```python
data['ram'].fillna(0, inplace=True)
data.head()
```

## 2. Data Recovery

```python
extracted_ram = data['Title'].str.extract(r'(\d+)\s*GB', expand=False).astype(float)

data.loc[data['ram'] == 0, 'ram'] = extracted_ram
```

```python
data['ram'  ].value_counts()
```

```python
data['Brand'] = data['Brand'].str.upper().str.strip()
```

```python
data['Brand'].value_counts()
```

```python
brand_map = {
    'NEWLENOVO': 'LENOVO',
    'NEW': 'LENOVO',
    'NEWDE.L.L': 'DELL',
    'ROG': 'ASUS',
    'TITAN': 'MSI'
}
data['Brand'] = data['Brand'].replace(brand_map)


def fix_brand_from_title(row):
    title = str(row['Title']).upper()
    brand = str(row['Brand']).upper()
    
    if brand in ['EXCALIBERPC', 'MAVARK', 'ME2', '[2025]', '[2026]']:
        if 'MSI' in title: return 'MSI'
        if 'DELL' in title: return 'DELL'
        if 'ASUS' in title: return 'ASUS'
        if 'LENOVO' in title: return 'LENOVO'
        if 'ACER' in title: return 'ACER'
        if 'APPLE' in title: return 'APPLE'
    return brand

data['Brand'] = data.apply(fix_brand_from_title, axis=1)
```

```python
data.head(10)
```

```python
data['Brand'].value_counts()
```

```python
data['operating_system'].value_counts()
```

```python
os_map = {
    'P:Windows 11 Pro': 'Windows 11 Pro',
    'Windows 11': 'Windows 11 Pro',
    'Windows 10 or 11': 'Windows 11 Pro',
    '-': 'Unknown'
}
data['operating_system'] = data['operating_system'].replace(os_map)

def recover_os(row):
    if row['operating_system'] == 'Unknown':
        title = str(row['Title']).upper()
        if 'WINDOWS 11' in title: return 'Windows 11 Pro'
        if 'WINDOWS 10' in title: return 'Windows 10 Pro'
        if 'MAC' in title or 'APPLE' in title: return 'Mac OS'
    return row['operating_system']

data['operating_system'] = data.apply(recover_os, axis=1)
```

```python
data['operating_system'].value_counts()
```

```python
data.dropna(subset=['Price'], inplace=True)
```
